```python
import pandas as pd
from google.colab import auth
import gspread
from google.auth import default

# Re-authenticate and authorize gspread client
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# Re-define SHEET_URL and WORKSHEET_NAME (from previous cells)
SHEET_URL = "https://docs.google.com/spreadsheets/d/1GFc6OXPSCmG2ivteUR4gdTyv47_9VobYubuMck46kXE/edit?usp=sharing"
WORKSHEET_NAME = "工作表1"

# Attempt to re-load df_restaurants
gsheets = gc.open_by_url(SHEET_URL)
sheets_data = gsheets.worksheet(WORKSHEET_NAME).get_all_values()

# Check if the sheet is empty (only header exists or no data)
if len(sheets_data) <= 1:
    print("Warning: Google Sheet is empty or only contains headers. Generating sample data for demonstration.")
    # Generate sample data if the sheet is empty
    sample_data = {
        '餐廳名稱': ['餐廳A', '餐廳B', '餐廳C', '餐廳D', '餐廳E', '餐廳F'],
        '價位': [150, 350, 200, 100, 400, 250],
        '距離': [500, 1200, 800, 300, 1500, 700],
        '是否營業': ['True', 'False', 'True', 'True', 'False', 'True'],
        '特色描述': ['好吃便宜', '高檔料理', '家庭餐廳', '快速午餐', '異國風味', '特色小吃']
    }
    df_restaurants = pd.DataFrame(sample_data)
else:
    df_restaurants = pd.DataFrame(sheets_data[1:], columns=sheets_data[0])

# User Input for Filtering (from previous cells)
max_price = 300
max_distance = 1000  # meters
open_now = True
sort_by = 'price'  # 'price' or 'distance'
sort_order = 'asc' # 'asc' for ascending, 'desc' for descending

# Convert columns to appropriate types and filter using correct column names
df_restaurants['價位'] = pd.to_numeric(df_restaurants['價位'], errors='coerce').fillna(0)
df_restaurants['距離'] = pd.to_numeric(df_restaurants['距離'], errors='coerce').fillna(0)
# '是否營業' might be string 'True'/'False' from sheets or bool from sample, convert robustly
df_restaurants['是否營業'] = df_restaurants['是否營業'].astype(str).str.lower().map({'true': True, 'false': False}).fillna(False)

filtered_restaurants = df_restaurants[
    (df_restaurants['價位'] <= max_price) &
    (df_restaurants['距離'] <= max_distance) &
    (df_restaurants['是否營業'] == open_now)
]

print(f"Filtered Restaurants (Top 5 Rows):")
print(filtered_restaurants.head())
```

```python
sort_column_map = {
    'price': '價位',
    'distance': '距離'
}

ascending_map = {
    'asc': True,
    'desc': False
}

sorted_restaurants = filtered_restaurants.sort_values(
    by=sort_column_map.get(sort_by, '價位'),  # Default to '價位' if sort_by is not recognized
    ascending=ascending_map.get(sort_order, True) # Default to ascending if sort_order is not recognized
)

print(f"\nSorted Restaurants by {sort_by} in {sort_order} order (Top 5 Rows):")
print(sorted_restaurants.head())
```

```python
print("\nRecommended Restaurants:")
if not sorted_restaurants.empty:
    for index, row in sorted_restaurants.iterrows():
        print(f"  Name: {row['餐廳名稱']}, Price: {row['價位']}, Distance: {row['距離']}m, Open Now: {row['是否營業']}, Description: {row['特色描述']}")
else:
    print("No restaurants found matching your criteria.")
```
